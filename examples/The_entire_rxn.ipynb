## (Example 1) Hydrogen exchange

```python
import numpy as np

atoms = ["H", "H", "H"]

def make_H_exchange_geometries(R_fixed=0.74, R_far=4.0, n_points=50):
    """
    Collinear hydrogen exchange path:

        H1 + H2-H3  ->  H1-H2 + H3

    H2 is fixed at origin.
    H1 approaches from the left.
    H3 leaves to the right.

    At the beginning:
        distance(H2,H3) ≈ R_fixed
        distance(H1,H2) ≈ R_far

    At the end:
        distance(H1,H2) ≈ R_fixed
        distance(H2,H3) ≈ R_far
    """

    geometries = []

    R_left_values = np.linspace(R_far, R_fixed, n_points)
    R_right_values = np.linspace(R_fixed, R_far, n_points)

    for R_left, R_right in zip(R_left_values, R_right_values):

        coords = np.array([
            [-R_left,  0.0, 0.0],   # incoming H
            [ 0.0,     0.0, 0.0],   # central H
            [ R_right, 0.0, 0.0],   # outgoing H
        ])

        structure = [
            (atom, tuple(map(float, xyz)))
            for atom, xyz in zip(atoms, coords)
        ]

        geometries.append(structure)

    return geometries


geometries = make_H_exchange_geometries(
    R_fixed=0.74,
    R_far=4.0,
    n_points=50,
)

for geometry in geometries:
    print(geometry)
```

```python
import numpy as np


def _bond_distance(geometry, i, j):
    """Distance between atoms i and j in a geometry list."""
    ri = np.array(geometry[i][1], dtype=float)
    rj = np.array(geometry[j][1], dtype=float)
    return np.linalg.norm(ri - rj)


def select_ts_centered_window(
    geometries,
    half_width=0.5,
    outside_stride=5,
    keep_endpoints=True,
    return_indices=True,
):
    """
    Select a TS-centered subset of geometries.

    Assumes a 3-atom H-exchange geometry:

        H1 --- H2 --- H3

    where the TS-like geometry is identified by

        distance(H1,H2) ≈ distance(H2,H3).

    The window is a square window in the 2D coordinate

        (R_left, R_right) = (d(H1,H2), d(H2,H3)).

    A geometry is inside the TS window if

        |R_left  - R_TS| <= half_width
        |R_right - R_TS| <= half_width

    All geometries inside the window are kept.
    Outside the window, only every `outside_stride`-th geometry is kept.

    Parameters
    ----------
    geometries:
        List of molecular geometries.

    half_width:
        Half-width of the square TS window in Angstrom.

    outside_stride:
        Downsampling factor outside the TS window.
        Example: outside_stride=5 keeps every 5th outside geometry.

    keep_endpoints:
        If True, always keep first and last geometries.

    return_indices:
        If True, return (selected_geometries, selected_indices, info).
        If False, return only selected_geometries.

    Returns
    -------
    selected_geometries:
        Reduced geometry list.

    selected_indices:
        Indices of selected geometries from the original list.

    info:
        Dictionary containing TS index, TS bond distance, and window data.
    """

    if len(geometries) == 0:
        if return_indices:
            return [], [], {}
        return []

    # ------------------------------------------------------------
    # 1. Compute reaction coordinates:
    #    R_left = d(H1,H2), R_right = d(H2,H3)
    # ------------------------------------------------------------

    R_left = np.array([
        _bond_distance(g, 0, 1)
        for g in geometries
    ])

    R_right = np.array([
        _bond_distance(g, 1, 2)
        for g in geometries
    ])

    # ------------------------------------------------------------
    # 2. Find TS-like middle geometry
    # ------------------------------------------------------------
    # For your path, this is where R_left ≈ R_right.
    # If n_points is even, there may not be an exactly symmetric point,
    # so we choose the closest one.
    # ------------------------------------------------------------

    ts_index = int(np.argmin(np.abs(R_left - R_right)))

    R_ts_left = R_left[ts_index]
    R_ts_right = R_right[ts_index]

    # Use the average as the TS center value.
    R_ts = 0.5 * (R_ts_left + R_ts_right)

    # ------------------------------------------------------------
    # 3. Define square window centered at (R_ts, R_ts)
    # ------------------------------------------------------------

    inside_window = (
        (np.abs(R_left - R_ts) <= half_width)
        &
        (np.abs(R_right - R_ts) <= half_width)
    )

    # ------------------------------------------------------------
    # 4. Keep all inside-window geometries,
    #    keep fewer outside-window geometries
    # ------------------------------------------------------------

    selected_indices = []

    for k in range(len(geometries)):
        if inside_window[k]:
            selected_indices.append(k)
        else:
            if outside_stride is not None and outside_stride > 0:
                if k % outside_stride == 0:
                    selected_indices.append(k)

    # Always keep endpoints if requested
    if keep_endpoints:
        selected_indices.append(0)
        selected_indices.append(len(geometries) - 1)

    # Remove duplicates and preserve order
    selected_indices = sorted(set(selected_indices))

    selected_geometries = [
        geometries[k]
        for k in selected_indices
    ]

    info = {
        "ts_index": ts_index,
        "R_ts_left": float(R_ts_left),
        "R_ts_right": float(R_ts_right),
        "R_ts_center": float(R_ts),
        "half_width": float(half_width),
        "outside_stride": outside_stride,
        "n_original": len(geometries),
        "n_selected": len(selected_geometries),
        "selected_indices": selected_indices,
    }

    if return_indices:
        return selected_geometries, selected_indices, info

    return selected_geometries

updated_geometries, selected_indices, info = select_ts_centered_window(
    geometries,
    half_width=0.4,
    outside_stride=8,
    keep_endpoints=True,
)

print(info)
print("Original number of geometries:", len(geometries))
print("Updated number of geometries:", len(updated_geometries))
print("Selected indices:", selected_indices)
```

```python
from collections import Counter
import numpy as np
import cirq
from pyLIQTR.PhaseEstimation.pe import PhaseEstimation
from openfermion import MolecularData, get_fermion_operator, count_qubits
from openfermionpyscf import run_pyscf


def _phase_from_qpe_bits(bits_measured_order):
    """
    bits_measured_order should be:
        [measure_{p-1}, measure_{p-2}, ..., measure_0]

    This matches the ordering used inside your PhaseEstimation.estimate_phase().
    """
    phase = 0.0
    for idx, bit in enumerate(bits_measured_order[::-1]):
        phase += int(bit) / (2 ** (idx + 1))
    return phase


def _phase_to_energy_signed(phase, ev_time):
    """
    Assumes U = exp(-i H t). Then:
        exp(-i E t) = exp(2πi phase)

    Therefore:
        E = -theta / t,
    where theta is the signed phase angle in (-π, π].
    """
    theta = 2.0 * np.pi * phase

    # unwrap from [0, 2π) to (-π, π]
    if theta > np.pi:
        theta -= 2.0 * np.pi

    return -theta / ev_time


def _make_pyliqtr_resource_circuit(qpe_circuit):
    """
    PyLIQTR resource estimation usually wants a unitary-style circuit.
    The semiclassical QPE circuit contains measurements and classical controls.

    For resource counting, we:
      1. remove measurements,
      2. strip classical controls but keep the underlying quantum gates.

    This does not change the intended gate-cost model of the feedback rotations.
    """
    resource_circuit = cirq.Circuit()

    for op in qpe_circuit.all_operations():

        if isinstance(op.gate, cirq.MeasurementGate):
            continue

        if isinstance(op, cirq.ClassicallyControlledOperation):
            op = op.without_classical_controls()

        resource_circuit.append(op)

    return resource_circuit


def qpe_ground_energy_from_geometry(
    geometry,
    basis="sto-3g",
    charge=0,
    multiplicity=1,
    precision_order=8,
    repetitions=200,
    trotter_number=1,
    trotter_order=1,
    ev_time=None,
    init_state=None,
    occupied_indices=None,
    active_indices=None,
    pick="lowest",
    run_fci_check=False,
):
    """
    Given one molecular geometry, build H, perform QPE, estimate energy,
    and return the QPE circuit for pyLIQTR resource estimation.

    Parameters
    ----------
    geometry:
        OpenFermion geometry format, e.g.
        [
            ("H", (0.0, 0.0, 0.0)),
            ("H", (0.0, 0.0, 0.7414)),
        ]

    pick:
        "lowest"      -> returns the lowest sampled QPE energy.
                         This is the ground-state-oriented option.
        "most_common" -> returns the most frequently observed eigenvalue.
                         This estimates the dominant eigencomponent of init_state.

    Returns
    -------
    result : dict
        result["energy_estimate"]        : QPE energy estimate
        result["qpe_circuit"]            : runnable semiclassical QPE circuit
        result["resource_circuit"]       : pyLIQTR-friendly circuit
        result["phase_estimate"]         : chosen phase estimate
        result["counts"]                 : sampled bitstring counts
        result["all_sampled_energies"]   : decoded sampled energies
        result["molecule"]               : OpenFermion MolecularData object
        result["mol_ham"]                : molecular Hamiltonian
        result["init_state"]             : occupation list used for QPE
    """

    # ------------------------------------------------------------
    # 1. Build molecular Hamiltonian from geometry
    # ------------------------------------------------------------
    molecule = MolecularData(
        geometry=geometry,
        basis=basis,
        charge=charge,
        multiplicity=multiplicity,
    )

    molecule = run_pyscf(
        molecule,
        run_scf=True,
        run_fci=run_fci_check,
    )

    mol_ham = molecule.get_molecular_hamiltonian(
        occupied_indices=occupied_indices,
        active_indices=active_indices,
    )

    fermion_ham = get_fermion_operator(mol_ham)
    n_spin_orbitals = count_qubits(fermion_ham)

    # ------------------------------------------------------------
    # 2. Choose initial occupation state
    # ------------------------------------------------------------
    if init_state is None:
        if occupied_indices is None:
            n_active_electrons = molecule.n_electrons
        else:
            n_active_electrons = molecule.n_electrons - 2 * len(occupied_indices)

        if n_active_electrons > n_spin_orbitals:
            raise ValueError(
                f"n_active_electrons={n_active_electrons} is larger than "
                f"n_spin_orbitals={n_spin_orbitals}. Check active space."
            )

        init_state = [1] * n_active_electrons + [0] * (n_spin_orbitals - n_active_electrons)

    if len(init_state) != n_spin_orbitals:
        raise ValueError(
            f"len(init_state)={len(init_state)} but Hamiltonian has "
            f"{n_spin_orbitals} spin orbitals/qubits."
        )

    # ------------------------------------------------------------
    # 3. Choose evolution time
    # ------------------------------------------------------------
    # Small default avoids phase wrapping for many small test Hamiltonians.
    # Larger ev_time gives better energy resolution but risks aliasing.
    if ev_time is None:
        ev_time = 0.25

    unitary_kwargs = {
        "trotterize": True,
        "mol_ham": mol_ham,
        "ev_time": ev_time,
        "trot_num": trotter_number,
        "trot_ord": trotter_order,
    }

    # ------------------------------------------------------------
    # 4. Build QPE circuit using your PhaseEstimation class
    # ------------------------------------------------------------
    pe = PhaseEstimation(
        precision_order=precision_order,
        init_state=init_state,
        phase_offset=0.0,
        include_classical_bits=True,
        kwargs=unitary_kwargs,
    )

    pe.generate_circuit()
    qpe_circuit = pe.pe_circuit

    # ------------------------------------------------------------
    # 5. Simulate QPE and decode measured phases
    # ------------------------------------------------------------
    simulator = cirq.Simulator()
    sim_result = simulator.run(qpe_circuit, repetitions=repetitions)

    measurement_order = [f"measure_{i}" for i in range(precision_order - 1, -1, -1)]

    sampled = []
    for shot in range(repetitions):
        bits = tuple(int(sim_result.measurements[key][shot, 0]) for key in measurement_order)
        phase = _phase_from_qpe_bits(bits)
        energy = _phase_to_energy_signed(phase, ev_time)
        sampled.append((bits, phase, energy))

    counts = Counter(bits for bits, _, _ in sampled)

    if pick == "most_common":
        chosen_bits = counts.most_common(1)[0][0]
        chosen_phase = _phase_from_qpe_bits(chosen_bits)
        chosen_energy = _phase_to_energy_signed(chosen_phase, ev_time)

    elif pick == "lowest":
        chosen_bits, chosen_phase, chosen_energy = min(sampled, key=lambda x: x[2])

    else:
        raise ValueError("pick must be either 'lowest' or 'most_common'.")

    # ------------------------------------------------------------
    # 6. Make circuit suitable for pyLIQTR resource estimation
    # ------------------------------------------------------------
    resource_circuit = _make_pyliqtr_resource_circuit(qpe_circuit)

    output = {
        "energy_estimate": chosen_energy,
        "phase_estimate": chosen_phase,
        "chosen_bits": chosen_bits,
        "counts": counts,
        "all_sampled_energies": sampled,
        "qpe_circuit": qpe_circuit,
        "resource_circuit": resource_circuit,
        "phase_estimation_object": pe,
        "molecule": molecule,
        "mol_ham": mol_ham,
        "init_state": init_state,
        "ev_time": ev_time,
    }

    if run_fci_check and hasattr(molecule, "fci_energy"):
        output["fci_energy_check"] = molecule.fci_energy

    return output
```
