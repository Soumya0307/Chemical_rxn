## W circuit builder example

```python
from dissipative_algorithms import hamiltonians
```

```python
from dissipative_algorithms import one_ham_with_ckt
```

```python
n_qubits = 5

interaction_operator = one_ham_with_ckt.HamiltonianBuilder.build_molecule(
    (
        ('H', (0, 0, 0)),
        ('H', (1, 0, 0)),
        ('H', (0, 1, 0)),
    ),
    multiplicity = 2,
    charge = 0
).interaction

one_ham_with_ckt.build_W_circuits_only(
    system_qubits = list(range(n_qubits)),
    H_int = interaction_operator,
    anc = 1,
    tau = 0.001,
    filter_terms = [0, 1],
    pairs = [2, 3],
    weights = [0.1] * n_qubits,
    N_applications = 5,
    T_filter = 0.05,
)
```
